```python
# 必要ライブラリーの読み込み
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings("ignore")
```

```python
# Penguinデータセットの読み込み
df = pd.read_csv("https://raw.githubusercontent.com/oshimalab/info_sci_2/refs/heads/main/data/penguins_3classes.csv")

# irisデータセットの中身を表示
df
```

```python
# くちばしの長さ（ミリメートル）と体重（グラム）のデータを取り出す
X = df[["bill_length", "body_mass"]].values

# くちばしの長さと体重の関係をプロットする
plt.scatter(X[:,0], X[:,1])
plt.show()
```

```python
# Scikit-learnからK-means法をインポート
from sklearn.cluster import KMeans

# クラスター数3,乱数シード1でクラスタリング
kmeans = KMeans(n_clusters=3, random_state=1).fit(X)

# 分類後のクラスター番号を表示
print(kmeans.labels_)

# クラスターの中心を表示
print(kmeans.cluster_centers_)
```

```python
# KMeansでの分類結果をプロット
plt.scatter(X[:, 0], X[:, 1], s=10, c=kmeans.labels_)

# 各クラスターの中心を赤丸でプロット
plt.scatter(kmeans.cluster_centers_[:,0], kmeans.cluster_centers_[:,1], s=40, c="red")
plt.show()
```

```python
# 正しい分類結果
sns.scatterplot(data=df, x="bill_length", y="body_mass", hue="species")
plt.show()
```

```python
# Scikit-learnから混合ガウス法をインポート
from sklearn.mixture import GaussianMixture

# クラスター数3, 乱数シード0でクラスタリング
gm_labels = GaussianMixture(n_components=3, random_state=0).fit_predict(X)

# 混合ガウス法での分類結果をプロット
plt.scatter(X[:, 0], X[:, 1], s=10, c=gm_labels)
plt.show()
```

```python
# 各特徴量の分布と相関の表示
sns.pairplot(df, hue = "species")
```
